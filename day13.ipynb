```python
def validate_rows(df, required_fields):
    valid_records = []
    errors = []

    for index, row in df.iterrows():
        row_errors = []

        # Check required fields
        for field in required_fields:
            if field not in df.columns:
                continue

            if pd.isna(row[field]) or str(row[field]).strip() == "":
                row_errors.append(f"{field} is missing")

        if row_errors:
            errors.append({
                "row_number": index,
                "reason": ", ".join(row_errors),
                **row.to_dict()
            })
        else:
            valid_records.append(row.to_dict())

    return valid_records, errors
```

```python
orders_valid, orders_errors = validate_rows(
    orders, ["order_id", "customer_id", "order_status"]
)

customers_valid, customers_errors = validate_rows(
    customers, ["customer_id", "customer_unique_id"]
)

products_valid, products_errors = validate_rows(
    products, ["product_id"]
)

order_items_valid, order_items_errors = validate_rows(
    order_items, ["order_id", "product_id", "price"]
)
```

```python
import pandas as pd


def validate_numeric(value):
    return pd.isna(value) or pd.notna(pd.to_numeric(value, errors="coerce"))

def validate_date(value):
    return pd.isna(value) or pd.notna(pd.to_datetime(value, errors="coerce"))
```

```python
# Orders - Date validation
orders["purchase_date_valid"] = orders["order_purchase_timestamp"].apply(validate_date)

orders["delivery_date_valid"] = orders["order_delivered_customer_date"].apply(validate_date)


# Order Items - Numeric and date validation
order_items["price_valid"] = order_items["price"].apply(validate_numeric)

order_items["freight_value_valid"] = order_items["freight_value"].apply(validate_numeric)

order_items["shipping_date_valid"] = order_items["shipping_limit_date"].apply(validate_date)


# Products - Numeric validation
products["weight_valid"] = products["product_weight_g"].apply(validate_numeric)

products["length_valid"] = products["product_length_cm"].apply(validate_numeric)


print("Data type validation completed!")
```

```python
print("ORDERS")
print(orders.filter(like="_valid").sum())

print("\nORDER ITEMS")
print(order_items.filter(like="_valid").sum())

print("\nPRODUCTS")
print(products.filter(like="_valid").sum())
```

```python
# two business rules:
#1. Product/order item price must be greater than 0.
#2. Order status must belong to an allowed list.

allowed_status = [
    "delivered",
    "shipped",
    "canceled",
    "unavailable",
    "invoiced",
    "processing",
    "created",
    "approved"
]
```

```python

zip_pattern = r"^\d{5}$"
id_pattern = r"^[a-fA-F0-9]{32}$"

def validate(value, pattern):
    return bool(re.fullmatch(pattern, str(value))) if pd.notna(value) else False
```

```python
customers["zip_valid"] = customers["customer_zip_code_prefix"].apply(lambda x: validate(x, zip_pattern))

sellers["zip_valid"] = sellers["seller_zip_code_prefix"].apply(lambda x: validate(x, zip_pattern))
```

```python
customers["id_valid"] = customers["customer_id"].apply(lambda x: validate(x, id_pattern))
orders["id_valid"] = orders["order_id"].apply(lambda x: validate(x, id_pattern))
products["id_valid"] = products["product_id"].apply(lambda x: validate(x, id_pattern))
order_items["id_valid"] = order_items["order_id"].apply(lambda x: validate(x, id_pattern))
sellers["id_valid"] = sellers["seller_id"].apply(lambda x: validate(x, id_pattern))
```

```python
print("Invalid customer records:", (~customers["zip_valid"] | ~customers["id_valid"]).sum())
print("Invalid seller records:", (~sellers["zip_valid"] | ~sellers["id_valid"]).sum())
print("Invalid orders:", (~orders["id_valid"]).sum())
print("Invalid products:", (~products["id_valid"]).sum())
print("Invalid order items:", (~order_items["id_valid"]).sum())
```

SchemaError – when required columns are missing.
FileFormatError – when a CSV file is missing, empty or cannot be read.

```python
class SchemaError(Exception):
    pass

class FileFormatError(Exception):
    pass
```

```python
def check_schema(df, required_columns):
    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        raise SchemaError(f"Missing columns: {missing}")

    print("Schema is valid!")
```

```python
try:
    check_schema(orders, ["order_id", "customer_id", "order_status"])

except SchemaError as e:
    print("Schema Error:", e)
```

```python
import os


def read_file(path):
    try:
        if not os.path.exists(path):
            raise FileFormatError("File does not exist")

        if os.path.getsize(path) == 0:
            raise FileFormatError("File is empty")

        return pd.read_csv(path)

    except pd.errors.EmptyDataError:
        raise FileFormatError("CSV contains no data")

    except pd.errors.ParserError:
        raise FileFormatError("Invalid CSV format")
```

```python
try:
    test_df = read_file("data/orders.csv")

except FileFormatError as e:
    print("File Error:", e)
```

```python
import os

import pandas as pd


def safe_read_csv(path):
    try:
        if not os.path.exists(path):
            raise FileFormatError("File not found")

        if os.path.getsize(path) == 0:
            raise FileFormatError("File is empty")

        return pd.read_csv(path)

    except (FileFormatError, pd.errors.EmptyDataError, pd.errors.ParserError) as e:
        print("Error:", e)
        return None
```

```python
def validation_summary(df, valid_records, rejected_records, filename):

    total = len(df)
    valid = len(valid_records)
    rejected = len(rejected_records)

    print("\nFile:", filename)
    print("Total rows:", total)
    print("Valid rows:", valid)
    print("Rejected rows:", rejected)

    if total > 0:
        print("Rejection rate:", round(rejected / total * 100, 2), "%")
```

```python
validation_summary(orders, orders_valid, orders_errors, "orders.csv")

validation_summary(customers, customers_valid, customers_errors, "customers.csv")

validation_summary(products, products_valid, products_errors, "products.csv")

validation_summary(order_items, order_items_valid, order_items_errors, "order_items.csv")

```

```python
MAX_REJECTION_RATE = 0.20
```

```python
def check_rejection_limit(total, rejected):

    if total == 0:
        print("Cannot validate an empty file")
        return

    rejection_rate = rejected / total

    if rejection_rate > MAX_REJECTION_RATE:
        print("FAILED: Rejection rate exceeds 20%")
    else:
        print("PASSED: Rejection rate is within the limit")
```

```python
check_rejection_limit(len(orders), len(orders_errors))
check_rejection_limit(len(customers), len(customers_errors))
check_rejection_limit(len(products), len(products_errors))
check_rejection_limit(len(order_items), len(order_items_errors))
```
