```python
column_mapping = {
    'order_id': 'order_id',
    'customer_id': 'customer_id',
    'price': 'price',
    'freight_value': 'freight',
    'order_purchase_timestamp': 'order_timestamp'
}
```

```python
def map_columns(data, mapping):
    new_data = []

    for row in data:
        new_row = {}

        for old_name, new_name in mapping.items():
            if old_name in row:
                new_row[new_name] = row[old_name]

        new_data.append(new_row)

    return new_data
```

```python
sample_data = [
    {
        'order_id': '101',
        'customer_id': 'C001',
        'price': '500',
        'freight_value': '50',
        'order_purchase_timestamp': '2018-01-01'
    }
]

mapped_data = map_columns(sample_data, column_mapping)

print(mapped_data)
```

```python
#6
from datetime import datetime

def convert_data(data):
    for row in data:
        row['price'] = float(row['price'])
        row['freight_value'] = float(row['freight_value'])
        row['order_purchase_timestamp'] = datetime.strptime(
            row['order_purchase_timestamp'], '%Y-%m-%d %H:%M:%S'
        )

    return data
```

```python
sample_data = [
    {
        'price': '500.50',
        'freight_value': '50.25',
        'order_purchase_timestamp': '2018-01-01 10:30:00'
    }
]

result = convert_data(sample_data)

print(result)
print(type(result[0]['price']))
print(type(result[0]['order_purchase_timestamp']))
```
