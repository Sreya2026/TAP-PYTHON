```python
#4
import csv
from pathlib import Path

INPUT_FOLDER = Path(r"C:\Users\Sreya\Documents\Tap - Python\olist_dataset")

file_path = INPUT_FOLDER / "orders.csv"

with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.reader(file)

    header = next(reader)
    row_count = sum(1 for row in reader)

print("File Name:", file_path.name)
print("Total Number of Rows:", row_count)
```

```python
#6
file_path = INPUT_FOLDER / "orders.csv"

with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)
    
    missing_count = {column: 0 for column in reader.fieldnames}
    
    for row in reader:
        for column in reader.fieldnames:
            if row[column] is None or row[column].strip() == "":
                missing_count[column] += 1

print("Missing Values Per Column:")

for column, count in missing_count.items():
    print(column, ":", count)
```

```python
#7
csv_files = list(INPUT_FOLDER.glob("*.csv"))

for file_path in csv_files:
    with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
        reader = csv.reader(file)
        
        header = next(reader, [])
        row_count = sum(1 for row in reader)
        
        print("\nFile Name:", file_path.name)
        print("Number of Columns:", len(header))
        print("Number of Rows:", row_count)
        print("Column Names:", header)
```
